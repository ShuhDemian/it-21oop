1. Основи синтаксису та індентація
Python відрізняється від C++ або Java тим, що не використовує фігурні дужки {} для позначення блоків коду. Замість цього використовується індентація (відступи).


```python
if True:
    print("Цей рядок має відступ (індентацію)")
    print("Він належить до блоку if")
print("Цей рядок вже поза блоком if")
```

2-3. Змінні та динамічна типізація
У Python не потрібно оголошувати тип змінної перед її використанням. Тип визначається автоматично (динамічно) в момент присвоєння значення.

```python
# Створення змінних
x = 10          # int (ціле)
print(f"x = {x}, тип: {type(x)}")

x = "Hi"     # Тепер x став str рядком - це динамічна типізація
print(f"x = {x}, тип: {type(x)}")

a, b, c = 1, 2.5, "Python" # Множинне присвоєння
```

4. Базові операції та приведення типів
Приклад розв'язання задачі: Обчислення площі кола та переведення результату в рядок.

```python
import math

радіус_str = "10"  # Вхідні дані, введені, як рядок
радіус = int(радіус_str)  # Приведення до int

area = math.pi * (радіус ** 2) # Операції: множення та піднесення до степеня
print("Площа (float):", area)

area_int = int(area) # Відкидання дробової частини
print("Площа (int):", area_int)
```

5. Робота зі списками (List)
Списки — це впорядковані змінні колекції об'єктів.

```python
# Створення списку
фрукти = ["Яблуко", "Банан", "Вишня"]

# Додавання елемента
фрукти.append("Апельсин")

# Зчитування (індексація починається з 0)
print(f"Перший елемент: {фрукти[0]}")
print(f"Останній елемент: {фрукти[-1]}")

# Зміна елемента
фрукти[1] = "Груша"
print(фрукти)
```

6. Основні оператори
Приклад використання арифметичних, логічних операторів та операторів порівняння.

```python
номер = 11

# Логічні оператори (and, or, not) та порівняння (>, <, ==)
if номер > 10 and номер % 2 != 0:
    print(f"{номер} є числом більше 10 і воно непарне.")
else:
    print("Умова не виконана.")
```

7-8. Робота зі стрічками (Strings) та форматування
Рядок у Python поводиться подібно до списку символів.

```python
текст = "Python Programming"

# Стрічка як список (Slicing)
print(текст[0])      # P
print(текст[0:6])    # Python

# Форматування
імя = "Олег"
вік = 20

# Метод 1: f-strings
print(f"Студент {імя}, вік: {вік}")

# Метод 2: .format()
print("Студент {}, вік: {}".format(імя, вік))
```

9. Словники (Dictionaries)
Словники зберігають дані у форматі ключ: значення. Вони оптимізовані для швидкого пошуку.

```python
# Створення словника
студент = {
    "імя": "Володимир",
    "курс": 3,
    "бал": 85.5
}

# Зчитування за ключем
print(f"Студент: {студент['імя']}")

# Додавання нового ключа
студент["спеціальність"] = "ІТ"

# Зміна значення
студент["курс"] = 4
print(студент)
```

## Відповіді на запитання для самоконтролю

1. Що таке індентація? Для чого вона використовується в Python 3?
Індентація — це відступ (пробіли або табуляція) на початку рядка коду. В Python вона критично важлива, оскільки визначає межі блоків коду (тіло циклів, функцій, умовних операторів). Неправильна індентація призведе до помилки IndentationError.

2. Як задати змінну? Як задати її тип? Зміна типу та множинне присвоєння.

Змінна задається оператором присвоєння =, наприклад: x = 5.

Тип задавати не потрібно, Python визначає його автоматично (динамічна типізація).

Так, тип змінної можна змінити, просто присвоївши їй значення іншого типу (наприклад, після x = 5 написати x = "text").

Так, можна задати значення декільком змінним: x, y, z = 10, 20, 30.

5. Що таке список (list)? Основні оператори.
Список — це змінна впорядкована колекція об'єктів довільних типів.
Основні методи та оператори:

len(list) — довжина списку.

+ — конкатенація (об'єднання) двох списків.

8. Що таке словники? Як задавати та зчитувати значення?
Словник (Dictionary) — це невпорядкована (до версії 3.7) колекція пар ключ: значення. Ключі мають бути унікальними та незмінними (хешованими).

Задати: my_dict = {"key": "value"}.

Зчитати: val = my_dict["key"] або безпечно val = my_dict.get("key").

9. Різниця між списками та словниками. Універсальність та наслідування.

Різниця: Списки індексуються цілими числами (0, 1, 2...) і зберігають порядок. Словники індексуються ключами (рядками, числами, кортежами) і оптимізовані для пошуку по ключу, а не за позицією.

Універсальність: Словник є більш універсальним для моделювання реальних об'єктів (де є властивості та значення), тоді як список кращий для послідовностей однотипних даних.

Наслідування: Прямого наслідування (в об'єктно-орієнтованому сенсі) одного від іншого в стандартній бібліотеці немає, це різні структури даних. Однак, словник можна представити як список кортежів [(key1, val1), (key2, val2)]. Створити словник зі списку пар можна так: dict(list_of_pairs).
