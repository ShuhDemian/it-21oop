```python
# Завдання 1
# Вводимо числа
x = float(input("Введіть x: "))
y = float(input("Введіть y: "))
z = float(input("Введіть z: "))

# Знаходимо максимум і мінімум за допомогою вбудованих функцій
ma = max(x, y, z)
mi = min(x, y, z)

# Рахуємо за формулою: (максимум у квадраті) - (2 в степені x * мінімум)
u = (ma ** 2) - (2 ** x) * mi

print("Відповідь u =", u)
```

```python
# Завдання 2
x = float(input("Введіть x: "))

if x <= 7:
    # Якщо x менше або дорівнює 7
    f = 3 * x - 9
else:
    # Якщо x більше 7
    f = 1 / (x ** 2 - 4)

print("F(x) =", f)
```
