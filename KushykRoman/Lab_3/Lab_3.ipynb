## Хід роботи
1. Суть структурного програмування
Структурне програмування — це підхід, при якому будь-яка програма будується з трьох базових керуючих структур:
•	Слідування: інструкції виконуються послідовно, одна за одною.
•	Розгалуження (умови): виконання певного блоку коду залежить від істинності умови.
•	Цикл: багаторазове повторення блоку коду.
У Python структурність підтримується на рівні синтаксису: замість фігурних дужок {} для виділення блоків коду використовуються відступи (зазвичай 4 пробіли). Це автоматично робить код читабельним і структурованим.
2. Використання умовних операторів
У Python для розгалужень використовується конструкція if, elif (скорочення від else if) та else. Дужки навколо умови ставити не обов'язково, а після умови обов'язково ставиться двокрапка :.


```python
x = 10
if x > 0:
    print("Число додатне")
elif x == 0:
    print("Число дорівнює нулю")
else:
    print("Число від'ємне")
```

3. Оператори булевої логіки в Python 3
На відміну від багатьох інших мов, де використовуються символи &&, || та !, у Python булеві оператори пишуться словами, що робить код схожим на звичайну англійську мову:
•	and (логічне І): істинно, якщо обидві умови істинні.
•	or (логічне АБО): істинно, якщо хоча б одна умова істинна.
•	not (логічне НІ): змінює значення на протилежне.
4. Проектування складних умов
Складні умови створюються шляхом комбінування простих за допомогою булевих операторів. Для визначення пріоритету операцій можна використовувати круглі дужки.
Приклад перевірки того, чи входить число в певний діапазон і чи є воно парним:


```python
number = 14
if (number >= 10 and number <= 20) and not (number % 2 != 0):
    print("Число парне і знаходиться в діапазоні від 10 до 20")
    
# У Python також можна писати подвійні нерівності (як в математиці):
if 10 <= number <= 20 and number % 2 == 0:
    print("Цей запис працює так само, але він значно елегантніший")
```

5. Методи роботи з циклами та їх доцільність
У Python є два основних типи циклів: while та for.
•	Цикл while (цикл із передумовою):
Використовується тоді, коли кількість ітерацій заздалегідь невідома, і цикл має продовжуватися, поки виконується певна умова.


```python
k = 1
# Наприклад, пошук найменшого степеня (як у задачах, де m^k > n)
while 2**k <= 100:
    k += 1
print("Шуканий степінь:", k)
```

•	Цикл for (цикл ітерації):
Використовується тоді, коли відома точна кількість повторень або коли потрібно пройтися по елементах колекції (списку, рядку тощо). Найчастіше використовується з функцією range().


```python
# Обчислення добутку або суми ряду чисел
product = 1
for i in range(1, 6): # Цикл від 1 до 5 включно
    if i % 2 != 0:    # Якщо число непарне
        product *= i
print("Добуток непарних чисел від 1 до 5:", product)
```

Для керування циклами також використовуються оператори break (дострокове завершення циклу) та continue (перехід до наступної ітерації).

## Відповіді на запитання для самоконтролю
1. Дайте визначення поняттю “структурне програмування”.

Структурне програмування — це методологія розробки програмного забезпечення, яка базується на представленні програми у вигляді ієрархічної структури блоків. Головна ідея полягає у відмові від хаотичних переходів (таких як goto) на користь трьох базових керуючих конструкцій:
•	Слідування (Sequence): послідовне виконання команд одна за одною.
•	Розгалуження (Selection): вибір шляху виконання залежно від умови (if, else).
•	Повторення (Iteration): багаторазове виконання блоку коду (цикли).
Цей підхід робить код більш читабельним, логічним і легшим для налагодження.

2. Який синтаксис умовних операторів у Python 3?
У Python синтаксис умовних операторів базується на ключових словах if, elif та else. Особливістю є використання відступів (indentation) для виділення блоків коду та двокрапки (:) після умови. Дужки навколо умови не є обов'язковими.
Загальний вигляд:
if умова_1:
     тіло блоку 1
elif умова_2:
    тіло блоку 2 (необов'язкова частина)
else:
    тіло блоку інакше (необов'язкова частина)
    Приклад (повна конструкція):

```python
x = 10
if x > 0:
    print("Додатне")
elif x == 0:
    print("Нуль")
else:
    print("Від'ємне")
```

Тернарний оператор (однорядковий запис):
status = "Allow" if x > 18 else "Deny"

3. Що таке булеві оператори?
Булеві (логічні) оператори — це оператори, які використовуються для побудови складних логічних виразів, результатом обчислення яких є істина (True) або хибність (False).
У Python використовуються такі ключові слова:
•	and (логічне І): повертає True, тільки якщо обидва операнди істинні.
•	or (логічне АБО): повертає True, якщо хоча б один з операндів істинний.
•	not (логічне НІ): інвертує значення (змінює True на False і навпаки).
4. Логічні оператори для списків та різниця між “==” та “is”.
Логічні оператори для списків:
До списків (list) часто застосовують оператори входження:
•	in: перевіряє, чи міститься елемент у списку (наприклад, if 5 in my_list:).
•	not in: перевіряє відсутність елемента.
Також списки підтримують булеву оцінку: порожній список [] оцінюється як False, непорожній — як True.
Різниця між == та is:
•	== (рівність значень): перевіряє, чи значення двох об'єктів однакові.
o	Приклад: [1, 2] == [1, 2] поверне True.
•	is (ідентичність об'єкта): перевіряє, чи посилаються обидві змінні на той самий об'єкт у пам'яті (чи мають вони однакову id).
o	Приклад: Якщо створити два окремих списки a = [1] і b = [1], то a is b поверне False, хоча їх значення рівні.
5. Що таке цикл? Що таке умова та тіло циклу?
Цикл — це алгоритмічна конструкція, яка дозволяє багаторазово виконувати певний набір інструкцій.
•	Умова циклу: це логічний вираз, який перевіряється перед (або після) кожною ітерацією. Якщо умова істинна (True), цикл продовжується; якщо хибна (False) — цикл завершується.
•	Тіло циклу: це блок коду, який виконується на кожній ітерації циклу, доки умова залишається істинною.
6. Способи організації циклів та їх доцільність.
У Python існують два основних типи циклів:
1.	Цикл for (цикл обходу ітерованого об'єкта):
o	Як працює: Проходить по елементах послідовності (список, рядок, range).
o	Коли доцільно: Коли відома кількість ітерацій заздалегідь або потрібно обробити кожен елемент колекції.
o	Приклад: for i in range(10): ...
2.	Цикл while (цикл з передумовою):
o	Як працює: Виконується доти, доки умова є істинною.
o	Коли доцільно: Коли точна кількість повторень невідома заздалегідь і залежить від зміни змінних усередині циклу (наприклад, пошук числа, введення даних користувачем).
o	Приклад: while x < 100: ...
7. Що таке оператори модифікації циклів?
Це спеціальні інструкції, які дозволяють змінити стандартний потік виконання циклу:
•	break: негайно перериває виконання циклу повністю (вихід із циклу). Використовується, наприклад, для дострокового виходу при знаходженні потрібного значення.
•	continue: перериває поточну ітерацію і переходить до початку наступної (до перевірки умови). Використовується, щоб пропустити певні елементи.
•	else (у циклах): специфічний оператор Python. Блок else виконується один раз після завершення циклу, тільки якщо цикл не був перерваний оператором break.
