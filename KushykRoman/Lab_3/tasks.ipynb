```python
import math

def find_closest_point():
    # Задаємо координати точок у вигляді кортежів (x, y)
    # Ви можете змінити ці значення для перевірки
    a = (0, 0)
    b = (3, 4)
    c = (1, 1)

    # Обчислюємо відстані від точки 'a' до 'b' та 'c'
    dist_ab = math.dist(a, b)
    dist_ac = math.dist(a, c)

    print(f"Відстань між a та b: {dist_ab:.2f}")
    print(f"Відстань між a та c: {dist_ac:.2f}")

    # Порівнюємо відстані
    if dist_ab < dist_ac:
        print("Точка b знаходиться ближче до точки a.")
    elif dist_ac < dist_ab:
        print("Точка c знаходиться ближче до точки a.")
    else:
        print("Точки b і c знаходяться на однаковій відстані від a.")

# Виклик функції
find_closest_point()
```

```python
def calculate_f(x):
    # Перевіряємо першу умову: x <= 7
    if x <= 7:
        result = -3 * x + 9
    # Якщо x > 7, виконується блок else
    else:
        result = 1 / (x - 7)
        
    return result

# Приклади використання:
# Перевіримо для x = 5 (повинна спрацювати перша гілка)
x1 = 5
print(f"При x = {x1}, F(x) = {calculate_f(x1)}")

# Перевіримо для x = 9 (повинна спрацювати друга гілка)
x2 = 9
print(f"При x = {x2}, F(x) = {calculate_f(x2)}")
```
