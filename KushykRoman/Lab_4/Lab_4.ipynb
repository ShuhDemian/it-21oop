## Хід роботи
Частина 1: Теоретичні відомості та виконання завдань
1. Синтаксис функцій у Python 3
Функція — це іменований блок коду, який виконує певну дію. У Python вона оголошується за допомогою ключового слова def.


```python
def скажи_привіт():
    print("Привіт, студенте!")

# Виклик функції
скажи_привіт()
```

2. Параметри та аргументи
Це два поняття, які часто плутають, але різниця є,і вона важлива:
•	Параметри — це змінні, перелічені в дужках при оголошенні функції (ніби "місця" для даних).
•	Аргументи — це конкретні значення, які ми передаємо у функцію під час її виклику.

```python
# 'імя' та 'група' — це ПАРАМЕТРИ
def print_student_info(імя, група):
    print(f"Студент: {імя}, Група: {група}")

# 'Роман' та 'ІТ-21' — це АРГУМЕНТИ
print_student_info("Роман", "ІТ-21")
```

3. Робота з функціями як об'єктами (Callback)
У Python функції є об'єктами першого класу. Це означає, що функцію можна передати іншій функції як аргумент. Це називається callback (зворотний виклик).

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

# Функція вищого порядку, яка приймає іншу функцію (operation)
def calculate(a, b, операція):
    результат = операція(a, b)
    print(f"Результат операції: {результат}")

# Передаю функцію add як параметр (без дужок!)
calculate(5, 3, add)      # Виведе: 8
calculate(5, 3, multiply) # Виведе: 15
```

## Частина 2: Відповіді на запитання для самоконтролю

1. Назвіть переваги використання функцій.
•	Повторне використання коду (Reusability): Функцію можна написати один раз і викликати багато разів у різних місцях програми.
•	Структурування (Modularity): Розбиття великої задачі на менші підзадачі робить код читабельним.
•	Легке налагодження (Debugging): Простіше знайти помилку в одній невеликій функції, ніж у величезному полотні коду.
•	Зменшення дублювання: Дозволяє уникнути копіювання однакових блоків коду.
2. Як оголосити функцію в Python 3?
Функція оголошується за допомогою ключового слова def, після якого йде ім'я функції, круглі дужки з параметрами (або без них) та двокрапка. Тіло функції пишеться з відступом.
Приклад: def my_func(): pass
3. Як викликати функцію?
Щоб викликати функцію, потрібно написати її ім'я, а після нього — круглі дужки (). Якщо функція приймає параметри, в дужках вказуються аргументи.
Приклад: my_func()
4. Що таке аргументи функції? Що таке параметри функції?
•	Параметри: змінні, вказані в оголошенні функції (у заголовку def). Вони діють як локальні змінні всередині функції.
•	Аргументи: реальні значення, які передаються у функцію під час її виклику.
5. Як задати аргументам функції значення за замовчуванням?
Значення за замовчуванням вказуються у списку параметрів через знак дорівнює =. Якщо під час виклику аргумент не передано, використовується це значення.
Приклад:

```python
def привіт(імя="Користувач"):
    print(f"Привіт, {імя}")
```

6. Що таке функції першого класу (First-class functions)?
Це концепція, яка означає, що в даній мові програмування функції розглядаються як звичайні об'єкти (змінні). Їх можна:
•	Присвоювати змінним.
•	Зберігати в структурах даних (списках, словниках).
•	Передавати як аргументи іншим функціям.
•	Повертати з інших функцій.
7. Що таке функції вищого порядку (Higher-order functions)?
Це функції, які роблять хоча б одну з двох речей:
1.	Приймають одну або більше функцій як аргументи (приклад: map(), filter(), або наш приклад calculate вище).
2.	Повертають іншу функцію як результат.
