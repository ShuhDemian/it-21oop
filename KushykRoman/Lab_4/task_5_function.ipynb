```python
def sum_of_digits(n):
    
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

def find_numbers_with_sum(k, m):
    
    result = []
    
    for i in range(1, m + 1):
        if sum_of_digits(i) == k:
            result.append(i)
    return result


if __name__ == "__main__":
    try:
        
        k = int(input("Введіть бажану суму цифр (K): "))
        m = int(input("Введіть максимальне число (M): "))
        
        
        if k <= 0 or m <= 0:
            print("Будь ласка, введіть натуральні числа (більші за 0).")
        else:
            
            numbers = find_numbers_with_sum(k, m)
            
            
            if numbers:
                print(f"\nЧисла, які не більші за {m} і сума цифр яких дорівнює {k}:")
                print(", ".join(map(str, numbers)))
            else:
                print(f"\nНе знайдено жодного числа до {m}, сума цифр якого дорівнює {k}.")
                
    except ValueError:
        print("Помилка: введено некоректні дані. Будь ласка, введіть цілі числа.")
```
